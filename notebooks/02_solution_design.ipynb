**How to read this:**
- `pretrained_imagenet_weights_loaded` must be **True**.
- `pos_weight` ≈ negatives ÷ positives in train, so a missed fault costs about 4× a false alarm in the loss.
- The over-fit test must drive the loss close to 0 and reach accuracy 1.0. That proves the input, output and loss are wired correctly (the model *can* learn).
